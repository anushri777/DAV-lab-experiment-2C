```python
import pandas as pd
import os

# Read CSV
text_df = pd.read_csv(os.path.expanduser("~/Downloads/sample_billing.csv"))

# Read Excel
excel_df = pd.read_excel(os.path.expanduser("~/Downloads/sample_billing.xlsx"))

# Read data from the web
web_df = pd.read_csv("https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv")

# Display data
print(text_df.head())
print(excel_df.head())
print(web_df.head())

# Handle missing values
text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
web_df.dropna(inplace=True)

# Save processed data
text_df.to_csv("processed_text.csv", index=False)
excel_df.to_excel("processed_excel.xlsx", index=False)

print("Files processed successfully!")
```
